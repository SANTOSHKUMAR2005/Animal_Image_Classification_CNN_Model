```python
import os 
import numpy as np
import pandas as pd
import cv2
```

```python
dataset_path=r"C:\Users\sank2\Desktop\Deep Learning\CNN\animals"
batch_size=32;
data=[]
```

```python
animal_classes=[d for d in os.listdir(dataset_path) if os.path.isdir(os.path.join(dataset_path,d))]

for class_name in animal_classes :
    class_folder_path=os.path.join(dataset_path,class_name)
    img_list=os.listdir(class_folder_path)

    for img_name in img_list :
        img_path=os.path.join(class_folder_path,img_name)

        # Verify file is an image extension
        if not img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
            continue
              
        try:
            # Read in color mode (RGB/BGR)
            img = cv2.imread(img_path, cv2.IMREAD_COLOR)
            if img is None:
                continue
                
            # Resize image to match column grid exactly
            img_resized = cv2.resize(img, (32, 32))
            
            # Flatten 3D matrix (64x64x3) to 1D array (12,288 values)
            flattened_pixels = img_resized.flatten()
            
            # Pack label and values together
            row = np.append(class_name, flattened_pixels)
            data.append(row)
            
        except Exception as e:
            pass  # Silently skip corrupted images to clean up outputs
```

```python
columns=["level"]
for c in ["R","G","B"]:
    for i in range(32):
        for j in range(32):
            columns.append(f'{c}_{i}x{j}')
```

```python
df=pd.DataFrame(data, columns=columns)
df.shape
```
